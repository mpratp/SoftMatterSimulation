**Lists**

```python
# This is a list
energy_kcal = [-13.4, -2.7, 5.4, 42.1]
# I can determine its length (len())
energy_length = len(energy_kcal)

# print the list length
print('The length of this list is', energy_length)
```

```python
# Print the first element of the list (0)
print(energy_kcal[0])
```

```python
# Convert the second list element to kilojoules.
energy_kilojoules = energy_kcal[1] * 4.184
print(energy_kilojoules)
```

**Slices**

```python
# list with few elements of previous list (slice)
# new_list = list_name[start:end]
# counting starts at 0, end doesn't count the last number but the previous one
short_list = energy_kcal[0:2]
print(short_list)
```

```python
slice1 = energy_kcal[1:]
slice2 = energy_kcal[:3]
print('slice1 is', slice1)
print('slice2 is', slice2)
```

```python
print(energy_kcal)
energy_kcal[0:2] # no new variable, energy_kcal remains the same
print(energy_kcal)
```

**Loops**

```python
# do something to every element of the list
# for variable in list:
    # do thing using variable
for number in energy_kcal:
    kJ = number * 4.184
    print(kJ)
```

```python
# append function adds a new item to the end of an existing list
# start list outside of the loop
energy_kJ = []
for number in energy_kcal:
    kJ = number * 4.184
    energy_kJ.append(kJ)

print(energy_kJ)
```

**Logic statements**

```python
# evaluate a variable and do something if it has a particular value (if)
negative_energy_kJ = []

for number in energy_kJ:
    if number < 0:
        negative_energy_kJ.append(number)

print (negative_energy_kJ)
```

**other logic operators:**
equal to ==
not equal to !=
greater than >
less than <
greater than or equal to >=
less than or equal to <=
You can also use and, or, and not to check more than one condition.

```python
negative_numbers = []
for number in energy_kJ:
    if number < 0 or number == 0:
        negative_numbers.append(number)

print(negative_numbers)
```
